```python
import numpy as np
import matplotlib.pyplot as plt

users = [10, 50, 100]

small_runs = [
    [610, 490, 382],
    [364, 356, 345],
    [355, 351, 340]
]

medium_runs = [
    [400, 447, 459],
    [387, 387, 384],
    [392, 385, 383]
]

large_runs = [
    [651, 647, 703],
    [636, 646, 636],
    [633, 640, 636]
]

fig, ax = plt.subplots(figsize=(10,6))

# plot individual runs
for run in small_runs:
    ax.plot(users, run,
            color='C0',
            alpha=0.25,
            linewidth=1)

for run in medium_runs:
    ax.plot(users, run,
            color='C1',
            alpha=0.25,
            linewidth=1)

for run in large_runs:
    ax.plot(users, run,
            color='C2',
            alpha=0.25,
            linewidth=1)

# plot means
ax.plot(users,
        np.mean(small_runs, axis=0),
        color='C0',
        marker='o',
        linewidth=3,
        label='Small')

ax.plot(users,
        np.mean(medium_runs, axis=0),
        color='C1',
        marker='s',
        linewidth=3,
        label='Medium')

ax.plot(users,
        np.mean(large_runs, axis=0),
        color='C2',
        marker='^',
        linewidth=3,
        label='Large')

ax.set_title('Response Time vs Concurrent Users',
             fontsize=16,
             fontweight='bold')

ax.set_xlabel('Concurrent Users')
ax.set_ylabel('Average Response Time (ms)')

ax.grid(alpha=0.3)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.legend(frameon=False)

plt.tight_layout()
plt.savefig('response_time.png', dpi=600, bbox_inches='tight')
plt.show()
```

```python
users = [10,50,100]

small = [19.93,99.13,196.20]
medium = [20.93,90.80,182.00]
large = [13.80,64.63,123.90]

fig, ax = plt.subplots(figsize=(10,6))

ax.plot(users, small,
        marker='o',
        linewidth=3,
        markersize=8,
        label='Small')

ax.plot(users, medium,
        marker='s',
        linewidth=3,
        markersize=8,
        label='Medium')

ax.plot(users, large,
        marker='^',
        linewidth=3,
        markersize=8,
        label='Large')

for x,y in zip(users,small):
    ax.annotate(f"{y:.0f}",(x,y),xytext=(0,8),
                textcoords="offset points",
                ha='center')

ax.set_title("Throughput vs Concurrent Users",
             fontsize=16,
             fontweight='bold')

ax.set_xlabel("Concurrent Users")
ax.set_ylabel("Requests per Second")

ax.grid(alpha=0.3)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.legend(frameon=False)

plt.tight_layout()
plt.savefig("throughput.png", dpi=600)
plt.show()
```

```python
import matplotlib.pyplot as plt

users = ['400', '800', '1200', '1500']
errors = [0, 0, 4.05, 23.33]

fig, ax = plt.subplots(figsize=(9, 5))

bars = ax.bar(
    users,
    errors,
    width=0.7,
    edgecolor='black',
    linewidth=0.8
)

ax.set_title(
    'Error Rate vs Concurrent Users',
    fontsize=16,
    fontweight='bold',
    pad=15
)

ax.set_xlabel('Concurrent Users', fontsize=12)
ax.set_ylabel('Error Rate (%)', fontsize=12)

ax.set_ylim(0, 30)

ax.grid(
    axis='y',
    linestyle='--',
    alpha=0.4
)

for bar, value in zip(bars, errors):
    ax.annotate(
        f'{value:.2f}%',
        xy=(bar.get_x() + bar.get_width()/2, value),
        xytext=(0, 6),
        textcoords='offset points',
        ha='center',
        fontsize=10,
        fontweight='bold'
    )

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig('errors.png', dpi=600, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt

tests = [
    "Moderate Load",
    "High Load",
    "Peak Load"
]

values = [718,895,1000]

fig, ax = plt.subplots(figsize=(9,5))

bars = ax.barh(tests, values)

ax.set_title("Concurrent Lambda Executions",
             fontsize=16,
             fontweight='bold')

ax.set_xlabel("Concurrent Executions")

for bar in bars:
    ax.text(
        bar.get_width()+10,
        bar.get_y()+bar.get_height()/2,
        f"{int(bar.get_width())}",
        va='center'
    )

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig("concurrency.png", dpi=600)
plt.show()
```

```python
import matplotlib.pyplot as plt

time = [0,2,4,6,8,10,12,14]
users = [0,10,50,100,100,50,10,0]

fig, ax = plt.subplots(figsize=(11,5))

ax.step(time,
        users,
        where='post',
        linewidth=3)

ax.fill_between(
    time,
    users,
    step='post',
    alpha=0.2
)

ax.set_title(
    "Workload Profile Used During Performance Testing",
    fontsize=16,
    fontweight='bold'
)

ax.set_xlabel("Time")
ax.set_ylabel("Active Users")

ax.grid(alpha=0.3)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig("workload_profile.png", dpi=600)
plt.show()
```
